## Apa kegunaan kedua metode tersebut?

Metode tersebut merupakan metode _Root of Equation_, digunakan untuk mencari Root (atau $x$) yang membuat f($x$) = 0.

Misalkan $f(x) = x^2 - 4$. Kita harus evaluate bahwa $x = ?$ sehingga $f(x) = 0$. Kita tahu jawabannya adalah $-2$ atau $2$. Namun untuk rumus yang lebih kompleks, metode ini tidak bekerja dengan baik.

Untuk itu, kita memiliki 2 cara yaitu **Bisection** dan **Newton-Raphson**


### Bisection

Metode Bisection merupakan Metode Bracketing, yang berarti menggunakan rumus

$$
m = \frac{a + b}{2}
$$

Diulangi hingga $$|f(m)| < \text{toleransi error}$$

Perlu dicatat, apabila:

- Simbol A == Simbol M, maka A = M.
- Simbol B == Simbol M, maka B = M.

Simbol, adalah +, -, atau 0. Misal 5 simbolnya +, -10 simbolnya -, dan 0... ya 0.


#### Python Code


````python
import numpy as np

def my_bisection(f, a, b, tol):
    """
    Approximates a root, R, of f bounded
    by a and b, to within a tolerance
    | f(m) | < tol with m the midpoint
    between a and b Recursive Implementation
    """
    
    # Check if a and b bound a root 
    if np.sign(f(a)) == np.sign(f(b)):
        raise Exception("The scalars a and b do not bound a root.")
    
    # Get Midpoint
    m = (a + b) / 2
    
    if np.abs(f(m)) < tol:
        # Stopping Condition, m is the root.
        return m
    elif np.sign(f(a)) == np.sign(f(m)):
        # Case where m is an improvement on a. Make recursive call with a = m
        return my_bisection(f, m, b, tol)
    elif np.sign(f(b)) == np.sign(f(m)):
        # Case where m is an improvement on b. Make recursive call with b = m
        return my_bisection(f, a, m, tol)
````

#### Implementasi Kode


````python
# f = lambda x: x**2 - 2 # This is also the same as b (lambda x is just a shorthand for f(x))
def f(x):
    return x**2 - 2

r1 = my_bisection(f, 0, 2, 0.1)
print(f"r1 = {r1}")

r01 = my_bisection(f, 2, 1, 0.01)
print(f"r01 = {r01}")

print(f"f(r1) = {f(r1)}")
print(f"f(r01) = {f(r01)}")
````

Melihat kembali ke Value, kita mendapatkan 2 hal disini. Pertama, parameter `my_bisection(f, a, b, tol)` memiliki arti sebagai berikut:

- `f` = Rumus yang akan dicari akarnya
- `a` = Batas Bawah
- `b` = Batas Atas
- `tol` = Nilai Tolerance, semakin rendah, semakin lebih presisi hasilnya, namun semakin lambat kalkulasinya. Hasil dari `f(f1, r01)` merepresentasikan hal tersebut (lebih dekat dengan 0 = lebih presisi).

### Newton-Raphson Method

Metode Newton-Raphson (kusingkat NR Method), adalah metode terbuka untuk mencari akar dengan cara mengierasikan rumus Newton dari $x_0$ hingga "error rate" lebih rendah daripada yang di tolerir.

Rumus dibalik NR sendiri adalah $$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}$$

Disini, mungkin kita notice bahwa Iterasi pertama adalah $x_1$, atau $x_{0+1}$, sehingga dibutuhkan tebakan awal yaitu $x_0$.


#### Python Code


````python
def newton_raphson(f, f_prime, x0, tol):
    """
    Approximates a root, R, of f using the
    Newton-Raphson method to within a tolerance
    | f(x) | < tol with x the current guess.
    Recursive Implementation
    """
    
    # Get next guess
    x1 = x0 - f(x0) / f_prime(x0)
    
    if np.abs(f(x1)) < tol:
        # Stopping Condition, x1 is the root.
        return x1
    else:
        # Make recursive call with new guess as the current guess.
        return newton_raphson(f, f_prime, x1, tol)
````

#### Implementasi kode


````python
f = lambda x: x**2 - 2
f_prime = lambda x: 2*x ## or f'(x) = 2x

nr_01 = newton_raphson(f, f_prime, 1.5, 1e-6) # 1e-6 = 0.000001 or 10^-6
print(f"Newton Raphson r1 = {nr_01}")
print(f"Square root of 2 = {np.sqrt(2)}")
````

### Pembuktian bahwa kedua metode menghasilkan hasil yang sama


Selama telah diketahui bahwa RoE $x^2 - 2$ adalah $\sqrt{2}$, kita dapat membuktikan bahwa hasilnya sama dengan Presisi Floating Point Number 5 Digit di Python.


````python
nr = newton_raphson(f, f_prime, 1.5, 1e-6)
bir = my_bisection(f, 0, 2, 1e-6)
sqrt = np.sqrt(2)

# Rounding to 5 Precision Digits
print(f"Newton Raphson r1 = {round(nr, 5)}")
print(f"Bisection r1 = {round(bir, 5)}") # type: ignore
print(f"Square root of 2 = {round(sqrt, 5)}")
````

### Metode yang lebih cepat

Untuk mengetahui speed dari masing-masing metode, rumus Python untuk Bisection dan Newton-Raphson saya modifikasi agar "lebih verbose". Sehingga kita dapat melihat angka yang didapatkan dimasing-masing iterasi. Serta berapa step yang dibutuhkan.

Agar konsisten, kita menggunakan rumus yang sama, $f(x) = x^2 - 2$


````python
bisection_iter, bisection_cycles = 0, 0
nr_iter, nr_cycles = 0, 0

def my_bisection_verbose(f, a, b, tol):
    global bisection_iter
    global bisection_cycles
    bisection_iter += 1
    """
    Approximates a root, R, of f bounded
    by a and b, to within a tolerance
    | f(m) | < tol with m the midpoint
    between a and b Recursive Implementation
    """
    
    # Check if a and b bound a root 
    if np.sign(f(a)) == np.sign(f(b)):
        raise Exception("The scalars a and b do not bound a root.")
    bisection_cycles += 1
    
    # Get Midpoint
    m = (a + b) / 2
    bisection_cycles += 1
    print(f"Iteration {bisection_iter: <3d} : a = {a :<10.6f}, b = {b :<10.6f}, f(a) = {f(a) :<10.6f}, f(b) = {f(b) :<10.6f}, m = {m :<10.6f}, f(m) = {f(m) :<10.6f}")
    
    if np.abs(f(m)) < tol:
        bisection_cycles += 1
        # Stopping Condition, m is the root.
        return m
    elif np.sign(f(a)) == np.sign(f(m)):
        bisection_cycles += 2
        # Case where m is an improvement on a. Make recursive call with a = m
        return my_bisection_verbose(f, m, b, tol)
    elif np.sign(f(b)) == np.sign(f(m)):
        bisection_cycles += 3
        # Case where m is an improvement on b. Make recursive call with b = m
        return my_bisection_verbose(f, a, m, tol)

def newton_raphson_verbose(f, f_prime, x0, tol):
    global nr_iter
    global nr_cycles
    nr_iter += 1
    """
    Approximates a root, R, of f using the
    Newton-Raphson method to within a tolerance
    | f(x) | < tol with x the current guess.
    Recursive Implementation
    """
    
    # Get next guess
    x1 = x0 - f(x0) / f_prime(x0)
    nr_cycles += 1
    
    if nr_iter < 10 or nr_iter > 2950:
        print(f"Iteration {nr_iter: <3d} : x0 = {x0 :<10.6f}, f(x0) = {f(x0) :<10.6f}, f'(x0) = {f_prime(x0) :<10.6f}, x1 = {x1 :<10.6f}, f(x1) = {f(x1) :<10.6f}")
    elif nr_iter == 10:
        print("...")
    
    if np.abs(f(x1)) < tol:
        # Stopping Condition, x1 is the root.
        nr_cycles += 1
        return x1
    else:
        # Make recursive call with new guess as the current guess.
        nr_cycles += 2
        return newton_raphson_verbose(f, f_prime, x1, tol)
    
print("Verbose Bisection Method:")
print(f"Bisection Result: {my_bisection_verbose(f, 0, 2, 1e-6) : .6f} // with {bisection_iter} iterations and {bisection_cycles} cycles")

print("\nVerbose Newton-Raphson Method:")
print(f"Newton-Raphson Result: {newton_raphson_verbose(f, f_prime, 1.5, 1e-6) : .6f} // with {nr_iter} iterations and {nr_cycles} cycles")
````

Dari yang kita bisa lihat, bahwa untuk Bisection Method membutuhkan 22 Iterasi dimana Newton-Raphson membutuhkan 3 Iterasi.

Selain itu, saya juga menambahkan verbosity untuk melihat instruksi yang harus dilalui algoritma masing-masing. Bisection menggunakan 98x Instruksi, dimana Newton-Raphson menggunakan 8x instruksi.

Do the Math $$(1 - \frac{8}{98}) \cdot 100 \approx 92 \%$$ dan Newton-Raphson ~92% lebih efisien daripada Bisection. (**_Note: Hitungan ini tidak sepenuhnya akurat dan bukan merupakan representasi beban CPU yang digunakan._**)


### Apakah Newton-Raphson selalu berhasil?

**Tidak**, ada beberapa rumus tertentu yang malah bisa bermasalah dengan kombinasi $x_0$ tertentu dan $f(x)$ tertentu.

Misalnya, apabila kita menggunakan $f(x) = x^3 - 2x + 2$ dengan $x_0 = 0$, maka Newton Raphson dapat terjebak dalam infinite loop, mengtrigger Recursion Error di Python.

Contohnya


````python
nr_iter, nr_cycles = 0, 0

import matplotlib.pyplot as plt

f = lambda x: x**3 - 2*x + 2
f_prime = lambda x: 3*x**2 - 2

try:
    newton_raphson_verbose(f, f_prime, 1, 1e-6)
except Exception as e:
    print(f"\nRecursion Limit Reached: {e}")
````

Namun isu ini tidak terjadi di Bisection, selama tidak terjadinya Skalar Error yang disebutkan sebelumnya


````python
bisection = my_bisection(f, -2, 2, 1e-6)
print(f"Bisection Result: {bisection:.12f}~ // with f(x) = {f(bisection):.12f}~")
````

### Kapan Bisection lebih baik?

Salah satunya dalam kasus rumus yang diturunkan justru menyebabkan error. Rumus seperti $$f(x) = |x|$$ apabila diturunkan menghasilkan $$\frac{x}{|x|}$$

Permasalahannya, kalkulasi tersebut tidak dapat menghasilkan 0. In other word... $$\frac{x}{|x|} = x$$ sehingga kalkulasi yang valid seharusnya $\frac{0}{|0|}$. Tapi ingat, $\frac{0}{|0|} = \text{Undefined}$, sehingga **tidak valid.**


````python
f = lambda x: abs(x)
f_prime = lambda x: 1 if x > 0 else -1 if x < 0 else 0 # This is an output of Copilot Model, but it also have the same meaning as x/abs(x)

for x in range(-10, 11):
    try:
        temp = newton_raphson(f, f_prime, x, 1e-6)
        print(f"Newton-Raphson Result for x = {x}: {temp} // f(x) = {f(temp)}")
    except Exception as e:
        print(f"Error Message: {e} at x = {x}")
````

````python
for x in range(-10, 11):
    try:
        temp = my_bisection(f, 0, 1*x, 1e-6)
        print(f"Bisection Result for x = {x}: {temp} // f(x) = {f(temp)}")
    except Exception as e:
        print(f"Error Message: {e} at x = {x}")
````

### Apakah Hasil Keduanya Sama?

Iya, tapi hingga dititik tertentu, apabila kita mengambil contoh dari Materi BINUSMAYA. Kita dapat menghitung selisihnya.


````python
f = lambda x: x**2 - 2
f_prime = lambda x: 2*x ## or f'(x) = 2x

bs = bisection = my_bisection(f, 0, 2, 1e-6)
nr = newton_raphson(f, f_prime, 1.5, 1e-6)

print(f"Bisection Result: {bs} // with f(x) = {f(bs)}")
print(f"Newton-Raphson Result: {nr} // with f(x) = {f(nr)}")
print(f"The Difference between Bisection and Newton-Raphson Result: {abs(bs - nr)} // or {abs(bs - nr) / abs(nr) * 100}%")
````

Hasilnya ada perbedaan tipis, namun apabila dibulatkan ke presisi digit tertentu, Angkanya dapat dikatakan "sama".
