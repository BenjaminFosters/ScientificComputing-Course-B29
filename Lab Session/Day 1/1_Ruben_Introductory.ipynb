Below here lies the data we're using! We're predicting **Final Term Score**, from **Mid Term Score**

```python
import numpy as np # Importing Numpy with np alias

# Class is a equiv of Struct.
class score:
    name = np.array(["Yudi", "Tono", "Naura", "Pras", "Yuli", "Yanto", "Simeon"])
    midterm = np.array([68, 75, 94, 82, 87, 90, 88])
    finterm = np.array([73, 74, 95, 82, 85, 93, 90])

for x in range(-1, len(score.name)):
    if (x == -1):
        # Format specifiers in C and Python are different. For example, in C, left alignment is `-` and right alignment is the default.
        # Python uses < for left alignment, and just like C, the default behavior is right alignment.
        print(f"{"Name":<7s} | {"Mid (x)":<8s} | {"Fin (y)":<8s}")
        for null in range(0, 30):
            print("-", end="")
        print("")
    else:
        print(f"{score.name[x]:<7s} | {score.midterm[x]:<8d} | {score.finterm[x]:<8d}")
```

# Calculating the RMSE, MSE, SSE

$$\text{SSE} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

$$\text{MSE} = \frac{1}{n} \sum^{n}_{i=1} (Y_{i} - \hat{Y}_{i}) ^ {2}$$

$$\text{RMSE}(y, \hat{y}) = \sqrt{\frac{\sum_{i=0}^{N - 1} (y_i - \hat{y}_i)^2}{N}}$$
